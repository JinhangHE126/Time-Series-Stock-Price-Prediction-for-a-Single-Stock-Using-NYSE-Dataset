```python
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio


df['date'] = pd.to_datetime(df['date'])


fig = go.Figure()

# add open line
fig.add_trace(go.Scatter(
    x=df['date'], y=df['open'],
    name='Open',
    mode='lines+markers',
    yaxis='y1',
    hoverinfo='all'
))

# add close line
fig.add_trace(go.Scatter(
    x=df['date'], y=df['close'],
    name='Close',
    mode='lines+markers',
    yaxis='y1',
    hoverinfo='all'
))

# add low line
fig.add_trace(go.Scatter(
    x=df['date'], y=df['low'],
    name='Low',
    mode='lines+markers',
    yaxis='y1',
    hoverinfo='all'
))

# add high line
fig.add_trace(go.Scatter(
    x=df['date'], y=df['high'],
    name='High',
    mode='lines+markers',
    yaxis='y1',
    hoverinfo='all'
))

# add volume line(right y-axis)
fig.add_trace(go.Scatter(
    x=df['date'], y=df['volume'],
    name='Volume',
    mode='lines+markers',
    yaxis='y2',
    hoverinfo='all'
))

# interactivity 
fig.update_layout(
    # title='AAPL 股票数据交互式图表',
    xaxis_title='日期',
    # 配置左侧Y轴（价格）
    yaxis=dict(
        title=dict(
            text='price (Open/Close/Low/High',
            font=dict(color='blue')
        ),
        tickfont=dict(color='blue'),
        side='left'
    ),
    # 配置右侧Y轴（成交量）
    yaxis2=dict(
        title=dict(
            text='成交量（Volume）',
            font=dict(color='red')
        ),
        tickfont=dict(color='red'),
        side='right',
        overlaying='y'
    ),
    # 图例放在图外右下角
    legend=dict(
        title='指标选择',
        itemclick='toggle',
        itemdoubleclick='toggleothers',
        x=1.02,
        y=-0.08,
        xanchor='left',
        yanchor='top'
    ),
    margin=dict(r=80, b=120),
    hovermode='x unified'
)

# ===================== 4. 显示图表 =====================
fig.show()
```
