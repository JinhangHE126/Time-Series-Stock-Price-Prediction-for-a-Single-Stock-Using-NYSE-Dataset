### Total Null values of each files:

```python
print(f'prices shape: {prices.shape}\n')

nulls = prices.isna().sum()

# Check if ANY column has more than 0 nulls
if nulls.any():
    print(f'prices.csv null values:\n{nulls[nulls > 0]}')
else:
    print('No null values in prices.csv')
print(f'\n')
print(f'\n')

print(f'prices_splits_adjusted shape: {prices_splits_adjusted.shape}\n')

nulls = prices_splits_adjusted.isna().sum()

# Check if ANY column has more than 0 nulls
if nulls.any():
    print(f'prices_splits_adjusted.csv null values:\n{nulls[nulls > 0]}')
else:
    print('No null values in prices_splits_adjusted.csv')
```

```python
print(f'fundamentals shape: {fundamentals.shape}\n ')

nulls = fundamentals.isna().sum()

# Check if ANY column has more than 0 nulls
if nulls.any():
    print(f'fundamentals.csv null values:\n{nulls[nulls > 0]}')
else:
    print('No null values in fundamentals.csv')

print(f'\n')
print(f'\n')
print(f'securities shape: {securities.shape}\n')
nulls = securities.isna().sum()

# Check if ANY column has more than 0 nulls
if nulls.any():
    print(f'securities.csv null values:\n{nulls[nulls > 0]}')
else:
    print('No null values in securities.csv')  
    
```
